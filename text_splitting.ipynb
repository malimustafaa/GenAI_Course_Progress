### WHY CONVERT_LISTS = TRUE


❗ But when your JSON starts with a LIST:

Like:

[
  {"id":1, "title":"A"},
  {"id":2, "title":"B"},
  {"id":3, "title":"C"}
]


There are no labels at the top.

Just a list of items.

The splitter gets confused because it tries to store something like:

chunk["?"] = item1


But…
there is no key. Nothing to attach it to.

📌 What convert_lists=True does (simple)

It creates fake labels like:

{
  "0": { ...first object... },
  "1": { ...second object... },
  "2": { ...third object... }
}


So now the splitter has:

a key

a value

And now it can split safely.

```python
from langchain_text_splitters import RecursiveJsonSplitter
json_splitter = RecursiveJsonSplitter(max_chunk_size=300)
json_chunks = json_splitter.split_json(json_data,convert_lists=True)
```

```python
json_chunks
```

```python
#first 3 chunks
```

```python
for chunk in json_chunks[:3]:
    print(chunk)
```

the json splitter can also create document type



```python
doc = json_splitter.create_documents(texts=[json_data],convert_lists=True)
for d in doc[:3]:
    print(d)
```
