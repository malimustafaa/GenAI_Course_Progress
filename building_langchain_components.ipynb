ChatPrompt Template - Telling our llm model its role, like act as an ai enginner and help me in coding, or act as a virtual assisitant for my company and cater to user queries

```python
from langchain_core.prompts import ChatPromptTemplate
```

```python
prompt = ChatPromptTemplate.from_messages(
    [
        ("system","you are an expert AI Engineer. Provide me answers based on questions"),
        ("user","{input}") #it means user is providing some input/questions, and u need to cater to those as an AI Engineer.

    ]
)
```

```python
print(prompt)
```

using prompt with my llm model


Chain basicially tell what the flow of the input is, first it goes to prompt then to llm model


```python
chain = prompt|llm #first it will go thru prompt, then llm
response = chain.invoke({"input":"Can you tell me about what is langsmith"})
print(response.content)
```

```python
type(response)
```

how we want to display our output message

```python
#stroutput parser
from langchain_core.output_parsers import StrOutputParser
output_parser = StrOutputParser()
chain = prompt|llm|output_parser
response = chain.invoke({"input":"Can you tell me about what is langsmith"})
print(response)
```
